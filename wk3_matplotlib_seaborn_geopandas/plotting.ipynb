```python
data['Date'] = pd.to_datetime(data['Date'])
```

```python
# creating a figure
plt.figure(figsize=(12,4))

# a simple line plot
plt.plot(data['Date'], data['ALUMINIUM'], linestyle='--', color='black')
plt.xlabel('Date')
plt.ylabel('Stock Price')
plt.title('Aluminium Stock Prices', fontsize=9, fontweight=300, loc='left', pad=15)

plt.show()
```

## Line plots

```python
# creating a figure
plt.figure(figsize=(12,4))

# a simple line plot
plt.plot(data['Date'], data['SILVER'], linestyle='--', color='black', label='Ag')
plt.plot(data['Date'], data['COPPER'], color='red', label='Cu')
plt.xlabel('Date')
plt.ylabel('Stock Price')
plt.title('Aluminium Stock Prices', fontsize=9, fontweight=300, loc='left', pad=15)
plt.legend()

plt.show()
```

### Adding subplots

```python
plt.subplots(nrows=1,ncols=2,figsize=(12,4))
```

```python
fig, ax = plt.subplots(nrows=1,ncols=2,figsize=(12,4))
ax[0].plot(data['Date'], data['SILVER'], linestyle='--', color='black', label='Ag')
ax[0].set_title('Silver Prices')
ax[0].set_ylabel('Prices')
ax[0].legend()
ax[1].plot(data['Date'], data['COPPER'], color='red', label='Cu')
ax[1].set_title('Copper Prices')
ax[1].set_ylabel('Prices')
ax[1].legend()

plt.show()
```

```python
fig,ax = plt.subplots(nrows=2,ncols=2,figsize=(12,4))
ax[0,0].plot(data['Date'], data['SILVER'], linestyle='--', color='grey', label='Ag')
ax[0,1].plot(data['Date'], data['COPPER'], linestyle='--', color='brown', label='Ag')
ax[1,0].plot(data['Date'], data['WHEAT'], linestyle='--', color='beige', label='Ag')
ax[1,1].plot(data['Date'], data['ALUMINIUM'], linestyle='--', color='blue', label='Ag')

plt.show()
```

### What are the total sales by lifestage

```python
# total sales by lifestage

lifestyle_sales = full_df.groupby('LIFESTAGE')[['TOT_SALES']].sum().sort_values(by='TOT_SALES').reset_index()

lifestyle_sales
```

```python
plt.figure(figsize=(6,5))
plt.bar(lifestyle_sales['LIFESTAGE'],lifestyle_sales['TOT_SALES'])
plt.tick_params(axis='x', labelrotation=45, labelsize=6)
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.barh(lifestyle_sales['LIFESTAGE'],lifestyle_sales['TOT_SALES'])
plt.title('Sales by Lifestage')

plt.show()
```

```python
sns.set_style('darkgrid')
```

```python
sns.barplot(lifestyle_sales, y='TOT_SALES', x='LIFESTAGE')
plt.title('Sales by Lifestage')
plt.show()
```
