### Step 4. How many products cost more than $10.00?

```python
# convert item_price to number

chipo['item_price'] = chipo['item_price'].replace('$', '', regex=True).astype(float)
```

```python
chipo.head()
```

```python
# check column dtypes to confirm

chipo.dtypes
```

```python
mask = chipo[chipo['item_price'] > 10.00]

chipo[mask].groupby('item_name')['item_price'].nunique()
```

### Step 5. What is the price of each item? 
###### print a data frame with only two columns item_name and item_price

```python
chipo.groupby('item_name')['item_price'].mean()
```

### Step 6. Sort by the name of the item

```python
chipo.groupby('item_name')['item_price'].mean().sort_index()
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
chipo.groupby('item_name')['quantity'].sum().max()
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
chipo.groupby('item_name')['quantity'].sum().loc['Veggie Salad Bowl']
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
# count of order_id where count of canned soda >1 per order_id

mask = chipo['quantity']>1
chipo_filtered = chipo[chipo['item_name'] == 'Canned Soda']

chipo_filtered[mask]['order_id'].nunique()
```
