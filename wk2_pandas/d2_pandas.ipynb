## Hierarchical Indexing/Multiindex
Useful when using an index on a single column is not sufficient to retreive the intended data e.g. moving from sales by country to sales by country, then by city, then by year(multiple dimensions)
- simplified idea: an index within an index

```python
# without hierarchical indexing

data = {
    "country": ["Kenya", "Kenya", "Uganda", "Uganda"],
    "city": ["Nairobi", "Mombasa", "Kampala", "Gulu"],
    "sales": [100, 80, 90, 40]
}

pd.DataFrame(data)

# country and city are regular columns, so some values are repeated
```

```python
# with hierarchical indexing

countries_n_cities = [
    ["Kenya", "Kenya", "Uganda", "Uganda"],
    ["Nairobi", "Mombasa", "Kampala", "Gulu"]
]

index = pd.MultiIndex.from_arrays(countries_n_cities, names=("country", "city"))

pd.DataFrame(
    {"sales": [100, 80, 90, 40]},
    index=index
)

# now there are 2 levels of the row index: country and city, where country groups multiple cities

# similar to merging cells in excel to show shared values?
# similar to using compound keys in sql

# q: is eliminating repetition that efficient?
```

```python
index = pd.MultiIndex.from_arrays(countries_n_cities, names=("country", "city"))

# index -> a variable containing the multiindex created
# pd.MultiIndex.from_arrays -> method to create a multiindex from an array (list of lists)
# names -> column headers for each of the indices
```

```python
# Sample sales data
data = {
    'Date': ['2022-01-01', '2022-01-01', '2022-01-02', '2022-01-02', '2022-01-03'],
    'Product': ['A', 'B', 'A', 'B', 'A'],
    'Region': ['North', 'South', 'North', 'South', 'North'],
    'Sales': [100, 150, 120, 80, 200],
    'Quantity': [2, 3, 1, 2, 4]
}

sales_df = pd.DataFrame(data)

# Convert 'Date' column to datetime type
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# Display the DataFrame
print("Sales DataFrame:")
print(sales_df)
```

```python
sales_df.index
```

```python
# manually set an index
# must save it in a variable!!! (inplace param OR new variable)

sales_df.set_index(['Date'], inplace=True)

# 'date' index column is separated(lower)
```
