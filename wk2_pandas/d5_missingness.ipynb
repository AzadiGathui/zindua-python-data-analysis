### Along rows

```python
data.notnull().all(axis=1)      # for each row, if all fields are notnull, return TRUE (TRUE -> complete row)
```

```python
data.notnull().all(axis=1).value_counts()   # shows how many rows are true vs False
```

```python
data.isnull().all(axis=1).value_counts()    # see all rows where ALL VALUES are missing (TRUE -> row missing all values)
```

```python
data.isnull().any(axis=1).value_counts()    # rows missing any value -> TRUE
```

```python
data.notnull().all(axis=0).value_counts()   # TRUE -> columns with complete values
```

```python
data[data.isnull().any(axis=1)]     # use missingness check as a mask -> returns rows missing any values
```

```python
data[~data.isnull().any(axis=1)]    # inverse of above
```
