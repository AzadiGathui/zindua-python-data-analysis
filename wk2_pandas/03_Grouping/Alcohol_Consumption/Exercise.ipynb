### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv). 

### Step 3. Assign it to a variable called drinks.

```python
data = pd.read_csv(r"D:\2-zindua\03-python-data-analysis\zindua-python-data-analysis\wk2_pandas\pandasdailypractice2-AzadiGathui\03_Grouping\Alcohol_Consumption\drinks.csv")

drinks = pd.DataFrame(data)
drinks.info()
```

```python
drinks.head()
```

### Step 4. Which continent drinks more beer on average?

```python
drinks.groupby('continent')['beer_servings'].max()
```

### Step 5. For each continent print the statistics for wine consumption.

```python
drinks['wine_servings'].describe()
```

### Step 6. Print the mean alcohol consumption per continent for every column

```python
drinks.groupby('continent')[['beer_servings', 'spirit_servings', 'wine_servings']].mean()
```

### Step 7. Print the median alcohol consumption per continent for every column

```python
drinks.groupby('continent')[['beer_servings', 'spirit_servings', 'wine_servings']].median()
```

### Step 8. Print the mean, min and max values for spirit consumption.
#### This time output a DataFrame

```python
spirit_consumption_df = drinks.groupby('continent')['spirit_servings'].agg(['min', 'max', 'mean'])

spirit_consumption_df
```
