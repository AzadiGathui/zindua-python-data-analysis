### Step 1. Import the necessary libraries

```python
import pandas as pd
import datetime as dt
import yfinance as yf
```

### Step 2. Create your time range (start and end variables). The start date should be 01/01/2015 and the end should today (whatever your today is)

```python
date_start = '2015-01-01'
date_end = dt.datetime.today().strftime('%Y-%m-%d')

date_end
```

### Step 3. Select the Apple, Tesla, Twitter, IBM, LinkedIn stocks symbols and assign them to a variable called stocks

```python
stocks = yf.download(['AAPL','NVDA','MSFT','TSLA'],date_start,date_end)
stocks
```

```python
stocks.info()
```

### Step 4. Read the data from google, assign to df and print it

```python
stocks = pd.DataFrame(stocks)
stocks
```

### Step 5.  What is the type of structure of df ?

```python
# a panel data structure?
```

### Step 6. Print all the Items axis values
#### To learn more about the Panel structure go to [documentation](http://pandas.pydata.org/pandas-docs/stable/dsintro.html#panel) 

```python
print(stocks.axes)
```

### Step 7. Good, now we know  the data avaiable. Create a dataFrame called vol, with the Volume values.

```python
vol = stocks['Volume']
vol.head()
```

### Step 8. Aggregate the data of Volume to weekly
#### Hint: Be careful to not sum data from the same week of 2015 and other years.

```python
vol_weekly = vol.resample('W').sum()
vol_weekly.head()
```

### Step 9. Find all the volume traded in the year of 2015

```python
vol_2015 = vol.groupby(vol.index.year).get_group(2015).sum()
vol_2015
```
