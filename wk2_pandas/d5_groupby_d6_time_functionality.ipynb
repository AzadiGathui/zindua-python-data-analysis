# Group by & Pivot tables
## Group by
- always group by sth categorical, repetitive
- computation on a numerical field against a categorical field
- i.e. `pd.groupby(")[aggregate_numerical]`

```python
full_df.columns
```

```python
# sales per lifestage
# is it repetitive?

full_df.groupby(['LIFESTAGE'])
```

```python
full_df.shape
```

```python
full_df['LIFESTAGE'].nunique()      # 7 != 264836
```

```python
full_df['LIFESTAGE'].unique()
```

```python
full_df.groupby('LIFESTAGE')['TOT_SALES'].sum()
```

```python
full_df.groupby('LIFESTAGE')[['TOT_SALES']].sum()
```

```python
full_df.groupby('LIFESTAGE')[['TOT_SALES']].sum().sort_values(by='TOT_SALES')
```

```python
full_df.groupby('LIFESTAGE')[['TOT_SALES']].sum().sort_values(by='TOT_SALES', ascending=False)
```

```python
full_df.groupby(['LIFESTAGE', 'PREMIUM_CUSTOMER'])[['TOT_SALES']].mean()    # order matters
```

```python
full_df.groupby(['LIFESTAGE', 'PREMIUM_CUSTOMER'])[['TOT_SALES', 'PROD_QTY']].mean()
```

```python
full_df.groupby(['LIFESTAGE']).agg({'TOT_SALES':'sum',
                                    'PROD_QTY':'mean'})
```

```python
full_df.groupby(['LIFESTAGE']).agg(
    total_sales = ('TOT_SALES','sum'),
    average_sales = ('TOT_SALES','mean')
)
```

## Pivot tables

```python
pd.pivot_table(full_df, values=['TOT_SALES','PROD_QTY'], index='LIFESTAGE',columns='PREMIUM_CUSTOMER', aggfunc='mean')
```
