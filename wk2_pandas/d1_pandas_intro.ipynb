### Creating DataFrame objects

```python
# from a list

data = [["Alice",25,"Kwale"], ["Bob",30,"Turkana"], ["Charlie",26,"Nairobi"]]
pd.DataFrame(data)
```

```python
# specifying column names

pd.DataFrame(data, columns=["Name", "Age", "City"])
```

```python
# from a dictionary of lists

data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 26],
        'City': ['Kwale', 'Turkana', 'Nairobi']}

pd.DataFrame(data)
```

```python
# from a dictionary of series objects

name_series = pd.Series(['Alice', 'Bob', 'Charlie'])
age_series = pd.Series([25, 30, 26])
city_series = pd.Series(['Kwale', 'Turkana', 'Nairobi'])

df = pd.DataFrame({"Name":name_series, "Age":age_series, "City":city_series})

df
```
