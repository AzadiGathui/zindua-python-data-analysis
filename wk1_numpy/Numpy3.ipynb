#### 33. How to get the dates of yesterday, today and tomorrow? (★☆☆)

````python
today = np.datetime64('today', 'D')
yesterday = today - np.timedelta64(1, 'D')
tomorrow  = today + np.timedelta64(1, 'D')

print(f'Today is {today}, yesterday was {yesterday} and tomorrow will be {tomorrow}.')
````

*dates*

- `np.datetime64` -> dates
- `np.timedelta64` -> date differences
- adding or subtracting works like normal math
- specify `D` to avoid time-of-day confusion

#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)

````python
# in place = without creating temporary arrays i.e. controlling memory

A = np.array([1,2,3,4,5], dtype=float)
B = np.array([5,4,3,2,1], dtype=float)

np.add(A,B,out=B)
np.divide(A,2,out=A)
np.negative(A,out=A)
np.multiply(A,B,out=A)
````

#### 38. Consider a generator function that generates 10 integers and use it to build an array (★☆☆)

````python
def gen_integers():
    for i in range(10):
        yield i #i.e. return the next int when asked

np.fromiter(gen_integers(), dtype=int)  #iterable here is the (for i in range...)
````

*generator functions*

- Uses yield instead of return
- Produces values one at a time
- Does not store everything in memory at once

#### 44. Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates (★★☆)

````python
arr44 = np.random.rand(10,2) *10
x = arr44[:,0]
y = arr44[:,1]
r = np.sqrt(x**2 + y**2)
theta = np.arctan2(y,x)
polar_coords = np.column_stack((r, theta))
polar_coords
````

*source: https://www.mathsisfun.com/polar-cartesian-coordinates.html*

- Using Cartesian Coordinates we mark a point by how far along and how far up it is


- Using Polar Coordinates we mark a point by how far away, and what angle it is
