# Concatenation
no. of input columns must match when concatenating along rows, except when flattening the output

```python
arr = np.array([1,2,3])
arr1 = np.array([[1,2,3]]) # must have multiple dimensions
arr2 = np.array([[4,5,6]])
```

```python
arr.shape
```

```python
arr1.shape
```

```python
arr2.shape
```

```python
# concat along rows

arr3 = np.concatenate((arr1,arr2))
arr3
```

```python
arr4 = np.concatenate((arr1,arr2), axis=None)   # flattened output
arr4
```

```python
# concat along columns

arr5 = np.concatenate((arr1,arr2), axis=1)
arr5
```

# Broadcasting
Broadcasting is how NumPy lets arrays of different shapes (multidimensional data) work together without copying data. It 'stretches' the smaller array to match the larger one.
When comparing shapes from right to left:
- Dimensions must be equal, or
- One of them must be 1, or
- One dimension can be missing
If not → ❌ error

```python
# 1d

arr = np.array([1,2,3])
arr + 10    # 10 is a scalar and is treated as [10,10,10]
```
