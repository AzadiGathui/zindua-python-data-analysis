#### 17. What is the result of the following expression? (★☆☆)
```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
# Each of these expressions has a different result

0 * np.nan          # result = nan
np.nan == np.nan    # can't be equal to itself, so result = False
np.inf > np.nan     # result = False
np.nan - np.nan     # result = nan
np.nan in set([np.nan])     # result = True
0.3 == 3 * 0.1      # result = False, bc the 0.3 float is an approximation, not exact
````
